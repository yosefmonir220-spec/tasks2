## 1. Introduction

Pandas is an open-source Python library used for **data manipulation, cleaning, exploration, and analysis**.

It is built around two core structures:

- **Series** — a one-dimensional labeled array
- **DataFrame** — a two-dimensional labeled table

Pandas is commonly used before machine-learning work for tasks such as data cleaning, exploratory analysis, feature preparation, and exporting processed data.

````python
import pandas as pd

print(f"Pandas version: {pd.__version__}")
````

## 2. Series

A **Series** is a one-dimensional labeled array.

You can think of it as:

- one column of a table
- a Python sequence with labels
- a structure where every value has an associated index

The default index is `0, 1, 2, ...`, but custom labels can also be used.

````python
scores = pd.Series([234, 542, 345, 685])
scores
````

### Custom indexes

The index does not have to be numeric. Meaningful labels can make data easier to access.

````python
scores = pd.Series(
    [234, 542, 345, 685],
    index=["student 1", "student 2", "student 3", "student 4"]
)

scores
````

````python
# Label-based access
print("student 1:", scores.loc["student 1"])

# Position-based access
print("third value:", scores.iloc[2])
````

### Updating Series values

- `.loc[]` accesses data by **label**
- `.iloc[]` accesses data by **position**

````python
scores.loc["student 2"] = 500
scores.iloc[2] = 400
scores.iloc[0] += 55

scores
````

### Filtering a Series

Boolean filtering works in a vectorized way, similar to NumPy.

````python
scores[scores >= 400]
````

### Creating a Series from a dictionary

Dictionary keys become the Series index.

````python
student_scores = {
    "student 1": 234,
    "student 2": 542,
    "student 3": 345,
    "student 4": 685
}

scores_from_dict = pd.Series(student_scores)
scores_from_dict
````

## 3. DataFrames

A **DataFrame** is a two-dimensional table with labeled rows and columns.

It is the main Pandas structure you will use for tabular data.

A useful mental model:

- rows → observations
- columns → variables/features
- index → row labels

A DataFrame column is itself a **Series**.

````python
students = pd.DataFrame({
    "Name": ["Marwan", "Ali", "Mahmoud", "Hager"],
    "Subteam": ["AI", "Navigation", "Hardware", "Navigation"],
    "Join_Date": [2025, 2025, 2024, 2025],
    "Score": [92, 85, 78, 95]
})

students
````

### Creating DataFrames in different ways

A DataFrame can be created from a dictionary, a list of lists, or with a custom index.

````python
# From a list of lists
df_list = pd.DataFrame(
    [[1, 2], [3, 4]],
    columns=["A", "B"]
)

df_list
````

````python
# With a custom index
df_custom = pd.DataFrame(
    [[1, 2], [3, 4]],
    columns=["A", "B"],
    index=["row_1", "row_2"]
)

df_custom
````

## 6. Accessing Columns and Rows

### Column access

A single column returns a Series:

```python
df["Score"]
```

Multiple columns return a DataFrame:

```python
df[["Name", "Score"]]
```

Dot notation such as `df.Score` is possible when the column name is a valid Python attribute, but bracket notation is generally safer and more consistent.

````python
students["Score"]
````

````python
students[["Name", "Subteam", "Score"]]
````

### `loc` vs `iloc`

**`loc`** is label-based.

```python
df.loc[row_label, column_label]
```

**`iloc`** is position-based.

```python
df.iloc[row_position, column_position]
```

A key difference is slicing:

- `loc` includes the ending label
- `iloc` excludes the ending position

````python
students.loc[0, "Score"]
````

````python
students.iloc[0, 2]
````

````python
students.loc[0:2, ["Name", "Score"]]
````

````python
students.iloc[0:3, 0:2]
````

### `at` and `iat`

When you need one specific cell:

- `.at[]` → label-based scalar access
- `.iat[]` → position-based scalar access

````python
print("at:", students.at[0, "Name"])
print("iat:", students.iat[0, 0])
````

## 8. Filtering Data

Boolean indexing returns rows where a condition is `True`.

Use:

- `&` for AND
- `|` for OR
- `~` for NOT

Put each condition inside parentheses.

````python
# Single condition
students[students["Score"] > 85]
````

````python
# AND
students[
    (students["Score"] > 80) &
    (students["Join_Date"] >= 2025)
]
````

````python
# OR
students[
    (students["Subteam"] == "AI") |
    (students["Subteam"] == "Hardware")
]
````

````python
# NOT
students[~(students["Subteam"] == "Navigation")]
````

### `isin()`

Useful when a column should match one of several possible values.

````python
students[students["Subteam"].isin(["AI", "Navigation"])]
````

### `between()`

`between()` checks whether values are inside an inclusive range.

````python
students[students["Score"].between(80, 95)]
````

### `query()`

`query()` provides a SQL-like way to write filters.

````python
students.query("Score > 85 and Join_Date >= 2025")
````

### `where()` and `mask()`

These are useful for conditional replacement.

- `where(condition)` keeps values where the condition is `True`
- `mask(condition)` replaces values where the condition is `True`

They are opposites.

````python
scores_demo = students["Score"].copy()

print("where():")
print(scores_demo.where(scores_demo >= 85))

print("\nmask():")
print(scores_demo.mask(scores_demo < 85, other=0))
````

## 9. Adding, Removing, and Transforming Columns

A column can be added directly.

You can create a column from:

- a constant
- another column
- a calculation involving several columns

````python
students_enhanced = students.copy()

students_enhanced["Passed"] = students_enhanced["Score"] >= 60
students_enhanced["Score_Percent"] = students_enhanced["Score"] / 100

students_enhanced
````

### `apply()`

`apply()` can apply a function to each value in a Series.

It is useful for readable custom transformations, although vectorized Pandas operations are usually preferred when an equivalent exists.

````python
students_enhanced["Performance"] = students_enhanced["Score"].apply(
    lambda x: "Excellent" if x >= 90
    else "Good" if x >= 80
    else "Needs Improvement"
)

students_enhanced[["Name", "Score", "Performance"]]
````

### `map()`

`map()` is convenient for element-wise replacement or transformation.

````python
students_enhanced["Subteam_Upper"] = students_enhanced["Subteam"].map(str.upper)

students_enhanced[["Name", "Subteam", "Subteam_Upper"]]
````

### `assign()`

`assign()` is useful when you want to create a chain of transformations without repeatedly mutating the original DataFrame.

````python
students_assigned = students.assign(
    Passed=students["Score"] >= 60,
    Score_Percent=students["Score"] / 100
)

students_assigned
````

### Removing columns

Use `drop()` to remove unwanted columns.

Prefer creating a new DataFrame when you want to keep the original unchanged.

````python
without_score = students.drop(columns=["Score"])
without_score
````

### `transform()`

`transform()` is especially useful with groups because it returns a result with the **same shape as the original data**.

For example, we can add the average score of each student's subteam to every corresponding row.

````python
students_with_group_mean = students.copy()

students_with_group_mean["Subteam_Mean_Score"] = (
    students_with_group_mean
    .groupby("Subteam")["Score"]
    .transform("mean")
)

students_with_group_mean[["Name", "Subteam", "Score", "Subteam_Mean_Score"]]
````

## 10. Data Cleaning

Data cleaning means finding and handling problems such as:

- missing values
- duplicate rows
- incorrect data types
- inconsistent text
- invalid or extreme values

A good cleaning workflow is:

**Detect → Decide → Handle → Validate**

### Missing values

First inspect missing values:

```python
df.isna().sum()
```

Then decide whether to drop or fill them.

````python
messy_students = pd.DataFrame({
    "Name": ["Ali", "Sara", "Omar", "Nour", "Nour"],
    "Age": [25, None, 22, 30, 30],
    "Score": [88, 95, None, 72, 72],
    "Subteam": ["AI", "Navigation", "AI", None, None]
})

messy_students
````

````python
messy_students.isna().sum()
````

### Dropping missing values

- `dropna()` removes rows containing missing values
- `dropna(subset=[...])` checks only selected columns
- `dropna(axis=1)` works on columns

````python
messy_students.dropna(subset=["Age"])
````

### Filling missing values

You can fill missing values with a constant or a statistic such as the mean.

````python
filled_students = messy_students.copy()

filled_students["Age"] = filled_students["Age"].fillna(
    filled_students["Age"].mean()
)

filled_students["Subteam"] = filled_students["Subteam"].fillna("Unknown")

filled_students
````

### Forward fill and interpolation

For ordered data, forward filling can carry the previous value forward.

Interpolation estimates missing numerical values between known values.

````python
time_data = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Value": [10.0, None, 30.0, None, 50.0]
})

time_data["Forward_Fill"] = time_data["Value"].ffill()
time_data["Interpolated"] = time_data["Value"].interpolate()

time_data
````

## 14. Outlier Detection

One common approach is the **IQR (Interquartile Range)** method.

1. Calculate Q1
2. Calculate Q3
3. `IQR = Q3 - Q1`
4. Values below `Q1 - 1.5 × IQR` or above `Q3 + 1.5 × IQR` are flagged

Outliers should not automatically be deleted. First determine whether they are errors or legitimate observations.

````python
prices = pd.Series([2.0, 2.5, 3.0, 3.2, 3.5, 4.0, 4.2, 25.0])

Q1 = prices.quantile(0.25)
Q3 = prices.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = prices[(prices < lower_bound) | (prices > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Outliers:")
print(outliers)
````

### Clipping values

If domain knowledge gives a valid range, `clip()` can constrain values to that range.

````python
clipped_prices = prices.clip(lower=0.5, upper=20)
clipped_prices
````

## 15. GroupBy and Aggregation

`groupby()` follows the **Split → Apply → Combine** idea:

1. **Split** rows into groups
2. **Apply** an aggregation or transformation
3. **Combine** the results

Common aggregations include `sum`, `mean`, `count`, `min`, and `max`.

````python
students.groupby("Subteam")["Score"].mean()
````

````python
students.groupby("Subteam").agg(
    Average_Score=("Score", "mean"),
    Maximum_Score=("Score", "max"),
    Student_Count=("Name", "count")
)
````

### `pivot_table()`

A pivot table summarizes data across two dimensions.

````python
sales = pd.DataFrame({
    "Day": ["Mon", "Mon", "Tue", "Tue", "Wed", "Wed"],
    "Type": ["Latte", "Tea", "Latte", "Tea", "Latte", "Tea"],
    "Revenue": [90, 40, 120, 55, 100, 60]
})

pd.pivot_table(
    sales,
    values="Revenue",
    index="Day",
    columns="Type",
    aggfunc="sum",
    fill_value=0
)
````

## 16. Merging DataFrames

Real datasets are often split across multiple tables.

`pd.merge()` combines DataFrames using matching keys.

Common join types:

- **inner** → only matching rows from both tables
- **left** → all rows from the left table
- **right** → all rows from the right table
- **outer** → all rows from both tables

````python
customers = pd.DataFrame({
    "CustomerID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Omar"]
})

orders = pd.DataFrame({
    "OrderID": [101, 102, 103, 104],
    "CustomerID": [1, 2, 2, 4],
    "Amount": [100, 150, 80, 200]
})

customers
````

````python
orders
````

````python
inner_join = pd.merge(
    orders,
    customers,
    on="CustomerID",
    how="inner"
)

inner_join
````

````python
left_join = pd.merge(
    orders,
    customers,
    on="CustomerID",
    how="left"
)

left_join
````

### `concat()`

`concat()` is useful for stacking DataFrames vertically or horizontally.

For example, when multiple DataFrames have the same columns, vertical concatenation combines their rows.

````python
part_1 = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Score": [88, 92]
})

part_2 = pd.DataFrame({
    "Name": ["Omar", "Nour"],
    "Score": [79, 95]
})

combined = pd.concat([part_1, part_2], ignore_index=True)
combined
````

## 17. Practical Data-Analysis Workflow

```text

Inspect

Clean

Select / Filter
  ↓
Transform
  ↓
Group / Merge
  ↓
Validate
  ↓

```

The important idea is that Pandas is not just about memorizing commands. The commands are tools used to move through this workflow.

## 19. Key Takeaways

- A **Series** is one-dimensional; a **DataFrame** is two-dimensional.
- Inspect your data before modifying it.
- Prefer `loc` for labels and `iloc` for positions.
- Use boolean conditions for flexible filtering.
- Use vectorized Pandas operations when possible.
- Treat data cleaning as a deliberate process: **detect → decide → handle → validate**.
- `groupby()` is one of the most important tools for summarizing data.
- `merge()` and `concat()` solve different table-combination problems.
- Keep the original DataFrame when possible by using `.copy()` before major transformations.
